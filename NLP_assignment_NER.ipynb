```python
import spacy

# Load spaCy model

lang_pipeline = 'en_core_web_sm' # with vectors

# lang_pipeline = 'en_core_web_md' #with vectors

try:

    # nlp = spacy.load(lang_pipeline)

    nlp = spacy.load(lang_pipeline)

except OSError:

    # from spacy.cli import download

    spacy.cli.download(lang_pipeline)

    nlp = spacy.load(lang_pipeline)

# Example text

text = "Adam will attend the Cyber Security conference in Lisbon on December 12, 2025."

# Process text

doc = nlp(text)

# Extract entities

person = None

event = None #'Cyber Conference'

location = None

date = None



for ent in doc.ents:

    print(ent.label_)

    if ent.label_ == "PERSON":

        person = ent.text

    elif ent.label_ == "GPE":

        location = ent.text

    elif ent.label_ == "ORG":

        event = ent.text + ' conference'

    elif ent.label_ == "DATE":

        date = ent.text



# Fill template

template = f"""

Conference Attendance Confirmation:

-----------------------------------

Attendee: {person}

Event: {event}

Location: {location}

Date: {date}

"""



print(template)
```

```python
import spacy

# Load the pre-trained SpaCy model
# nlp = spacy.load("en_core_web_sm")
lang_pipeline = 'en_core_web_sm' # with vectors

# lang_pipeline = 'en_core_web_md' #with vectors
try:

    # nlp = spacy.load(lang_pipeline)

    nlp = spacy.load(lang_pipeline)
except OSError:

    # from spacy.cli import download

    spacy.cli.download(lang_pipeline)

    nlp = spacy.load(lang_pipeline)

from spacy.tokens import Doc, Span
from spacy.matcher import Matcher

# Load the pre-trained SpaCy model
nlp = spacy.load("en_core_web_sm")

# Define the custom component
def extract_relations(doc):
    matcher = Matcher(nlp.vocab)
    # Define patterns for matching relations
    pattern = [
        {'DEP': 'nsubj'},
        {'DEP': 'aux', 'OP': '?'},
        {'DEP': 'ROOT'},
        {'DEP': 'det', 'OP': '?'},
        {'DEP': 'amod', 'OP': '*'},
        {'DEP': 'dobj'}
    ]
    matcher.add("relation_pattern", [pattern])
    matches = matcher(doc)

    relations = []
    for match_id, start, end in matches:
        span = doc[start:end]
        relations.append((span.text, span.root.dep_))
    return relations

# Register the custom component with SpaCy
Doc.set_extension("relations", getter=extract_relations, force=True)

# Sample text
text = "Apple is acquiring a U.K. startup."

# Process the text
doc = nlp(text)

# Extract and print relations
relations = doc._.relations
for relation in relations:
    print(relation)


from spacy import displacy

# Visualize the dependency parse tree
displacy.render(doc, style="dep", jupyter=True)
```
